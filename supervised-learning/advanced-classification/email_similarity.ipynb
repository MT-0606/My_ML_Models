# Email Similarity
**Objective:** Use ```scikit-learn```'s Naïve Bayes implementation on several different datasets. By reporting the accuracy of the classifier, we can find which datasets are harder to distinguish. For example, how difficult do you think it is to distinguish between emails about ice hockey and emails about soccer? How hard is it to tell the difference between emails about hockey and emails about tech? In this project, we'll find out exactly how difficult those two tasks are.

````python
from sklearn.datasets import fetch_20newsgroups
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer
````

## Explore the Data

````python
# 1. See the different categories.
emails = fetch_20newsgroups()
print(emails.target_names)
````

````python
# 2. See how effectively our Naïve Bayes classifier distinguishes between a baseball and ice hockey email.
emails = fetch_20newsgroups(categories=['rec.sport.baseball','rec.sport.hockey'])
````

````python
# 3. Print email #6 (index 5).
print(emails.data[5])
````

````python
# 4. Print the label of email #6.
print(emails.target[5])
````

**Verdict:** The email above was about ice hockey.

## Make the Training and Testing Sets

````python
# 5. Split the data into a training set.
training_emails = fetch_20newsgroups(categories=['rec.sport.baseball','rec.sport.hockey'],
                                     subset='train',
                                     shuffle=True,
                                     random_state=108)
````

````python
# 6. Split the data into a testing set.
testing_emails = fetch_20newsgroups(categories=['rec.sport.baseball','rec.sport.hockey'],
                                    subset='test',
                                    shuffle=True,
                                    random_state=108)
````

## Count Words

````python
# 7. Define a CountVectorizer object.
counter = CountVectorizer()
````

````python
# 8. Determine what possible words can exist in the emails.
counter.fit(testing_emails.data+training_emails.data)
````

````python
# 9. Make a list of word counts in our training set.
training_counts = counter.transform(training_emails.data)
````

````python
# 10. Make a list of word counts in our testing set.
testing_counts = counter.transform(testing_emails.data)
````

## Make a Naïve Bayes Classifier

````python
# 11. Make a Naïve Bayes classifier that we can train and test on.
classifier = MultinomialNB()
````

````python
# 12. Fit the classifier on our training set and the associated labels.
classifier.fit(training_counts, training_emails.target)
````

````python
# 13. Test the classifier.
print(classifier.score(testing_counts, testing_emails.target))
````

## Testing Other Datasets

````python
# 14. Determine how well the classifier distinguishes between different topics.
training_emails = fetch_20newsgroups(categories=['comp.sys.ibm.pc.hardware','rec.sport.hockey'],
                                     subset='train',
                                     shuffle=True,
                                     random_state=108)
testing_emails = fetch_20newsgroups(categories=['comp.sys.ibm.pc.hardware','rec.sport.hockey'],
                                    subset='test',
                                    shuffle=True,
                                    random_state=108)
classifier.fit(counter.transform(training_emails.data), training_emails.target)
print(classifier.score(counter.transform(testing_emails.data), testing_emails.target))
````
